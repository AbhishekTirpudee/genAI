```python
import requests
import time
import psutil
import threading
import subprocess
import pandas as pd
import gradio as gr
import matplotlib.pyplot as plt

OLLAMA_URL = "http://localhost:11434/api/generate"

print("All libraries imported successfully!")
```

```python
def get_gpu_memory():
    try:
        result = subprocess.check_output(
            ["nvidia-smi",
             "--query-gpu=memory.used",
             "--format=csv,noheader,nounits"],
            encoding="utf-8"
        )
        return sum(map(float, result.strip().split("\n")))
    except:
        return None


def benchmark_model(model, prompt):

    memory_samples = []
    gpu_samples = []

    baseline_ram = psutil.virtual_memory().used / (1024 ** 2)
    baseline_gpu = get_gpu_memory()

    running = True

    def monitor():
        while running:
            memory_samples.append(
                psutil.virtual_memory().used / (1024 ** 2)
            )

            gpu = get_gpu_memory()
            if gpu is not None:
                gpu_samples.append(gpu)

            time.sleep(0.1)

    monitor_thread = threading.Thread(target=monitor)
    monitor_thread.start()

    start_time = time.perf_counter()

    try:
        response = requests.post(
            OLLAMA_URL,
            json={
                "model": model,
                "prompt": prompt,
                "stream": False,
                "options": {
                    "temperature": 0.2
                }
            },
            timeout=600
        )

        response.raise_for_status()
        result = response.json()

        end_time = time.perf_counter()

    finally:
        running = False
        monitor_thread.join()

    response_time = end_time - start_time

    tokens = result.get("eval_count", 0)
    eval_duration = result.get("eval_duration", 0)

    tokens_per_second = (
        tokens / (eval_duration / 1e9)
        if eval_duration > 0 else 0
    )

    peak_ram = max(memory_samples) if memory_samples else baseline_ram
    ram_consumption = max(0, peak_ram - baseline_ram)

    peak_gpu = max(gpu_samples) if gpu_samples else None

    gpu_consumption = (
        max(0, peak_gpu - baseline_gpu)
        if peak_gpu is not None and baseline_gpu is not None
        else None
    )

    return {
        "Model": model,
        "Response Time (s)": round(response_time, 3),
        "Generated Tokens": tokens,
        "Tokens/sec": round(tokens_per_second, 2),
        "RAM Usage (MB)": round(ram_consumption, 2),
        "Response": result.get("response", "")
    }
```

```python
prompt = "Explain artificial intelligence in simple words."

result1 = benchmark_model(
    "qwen2.5:1.5b",
    prompt
)

print(result1)
```

```python
def compare_models(prompt, model1, model2):

    result1 = benchmark_model(model1, prompt)
    result2 = benchmark_model(model2, prompt)

    results = [result1, result2]

    df = pd.DataFrame([
        {k: v for k, v in r.items() if k != "Response"}
        for r in results
    ])

    # Performance charts
    fig, axes = plt.subplots(1, 3, figsize=(14, 4))

    metrics = [
        ("Response Time (s)", "Response Time"),
        ("Tokens/sec", "Tokens Per Second"),
        ("RAM Usage (MB)", "RAM Consumption")
    ]

    for ax, (column, title) in zip(axes, metrics):
        values = [r[column] for r in results]
        ax.bar([model1, model2], values)
        ax.set_title(title)
        ax.set_ylabel(column)
        ax.tick_params(axis="x", rotation=20)

    plt.tight_layout()

    comparison = f"""
    ## Model Comparison

    **Faster response:** {
        model1 if result1["Response Time (s)"] <
        result2["Response Time (s)"] else model2
    }

    **Higher token generation speed:** {
        model1 if result1["Tokens/sec"] >
        result2["Tokens/sec"] else model2
    }

    **Lower RAM increase:** {
        model1 if result1["RAM Usage (MB)"] <
        result2["RAM Usage (MB)"] else model2
    }
    """

    responses = (
        f"### {model1} Response\n{result1['Response']}\n\n"
        f"### {model2} Response\n{result2['Response']}"
    )

    return df, fig, comparison, responses
```

```python
with gr.Blocks(title="LLM Benchmarking Dashboard") as demo:

    gr.Markdown("# LLM Benchmarking Dashboard")
    gr.Markdown(
        "Compare two locally available LLM models "
        "based on response time, TPS and RAM consumption."
    )

    prompt = gr.Textbox(
        label="Enter your prompt",
        value="Explain machine learning in simple words.",
        lines=3
    )

    with gr.Row():
        model1 = gr.Dropdown(
            choices=["qwen2.5:1.5b", "llama3.2:1b"],
            value="qwen2.5:1.5b",
            label="Select Model 1"
        )

        model2 = gr.Dropdown(
            choices=["qwen2.5:1.5b", "llama3.2:1b"],
            value="llama3.2:1b",
            label="Select Model 2"
        )

    run_button = gr.Button(
        "Run Benchmark",
        variant="primary"
    )

    output_table = gr.Dataframe(
        label="Benchmark Results"
    )

    output_chart = gr.Plot(
        label="Performance Comparison"
    )

    output_comparison = gr.Markdown()

    output_responses = gr.Markdown()

    run_button.click(
        fn=compare_models,
        inputs=[prompt, model1, model2],
        outputs=[
            output_table,
            output_chart,
            output_comparison,
            output_responses
        ]
    )

demo.launch(share=True)
```
